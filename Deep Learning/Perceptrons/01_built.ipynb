```python
import pandas as pd 
import numpy as np 
```

```python
class Perceptron:

    # for binary classification 

    def __init__(self,learning_rate=0.01,n_iters= 1000):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.w_ = None # weights including bias 
        self.errors_ = []

    def weighted_sum(self,X):
        return np.dot(X,self.w_[1:] + self.w_[0])
    
    def fit(self,X,y):
        X = np.asanyarray(X,dtype=float)
        y = np.asarray(y)

        self.w_ = np.zeros(1 + X.shape[1])

        #training loop 
        for _ in range(self.n_iters):
            errors = 0
            for xi, target in zip(X,y):
                # preddict for one sample 
                y_pred = self.predict(xi)

                # compute update target - prediction
                update = self.learning_rate * (target - y_pred) 


                # update the feature 
                self.w_[1:] += update * xi 
                # update bias 
                self.w_[0] += update 

                # cout mistake if there was an update 
                errors += int(update != 0.0)

            self.errors_.append(errors)
        return self

    def predict(self,X):
        X = np.atleast_2d(X) # ensure shape 
        net_input = self.weighted_sum(X)
        return np.where(net_input >= 0.0,1,-1)
        


```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import  train_test_split
from sklearn.metrics  import accuracy_score

# load iris 
iris = load_iris()
X = iris.data 
y = iris.target

# keep ony 2 lasses 
mask = y < 2 
X_binary = X[mask]
y_binary = y[mask]

# Convert labels: 0 -> -1, 1 -> 1
y_binary = np.where(y_binary == 0, -1, 1)

# train test split 
X_train, X_test, y_train, y_test = train_test_split(
    X_binary, y_binary, test_size = 0.25, random_state = 42
    )

# train perceptron 
p = Perceptron(learning_rate=0.01,n_iters=10)
p.fit(X_train,y_train)

# predict 
y_pred = p.predict(X_test)

# evaluate 
accuracy_score(y_test,y_pred) * 100
```
