```python
import torch
import torch.nn as nn
```

```python
# creating a tensor
x = torch.tensor([1,2,3,4,5])
print(x)

# 2d tensors
y = torch.tensor([[1,2,3],
                  [4,5,6]])
print(y)

# creating tensors with zeros
z = torch.zeros([3,3])
print(z)

# with random
a = torch.randn(6,7)
print(a)

# with ones
torch.ones(2,3)
```

```python
#  Common operations
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print("\nAddition:", a + b)
print("Multiplication:", a * b)  # element-wise
print("Mean:", a.mean())
print("Sum:", a.sum())
```

```python
# Matrix multiplication (dot product)
A = torch.tensor([[1, 2],
                  [3, 4],
                  [5, 6]])  # 3x2

B = torch.tensor([[7, 8, 9],
                  [10, 11, 12]])  # 2x3


C = torch.matmul(A,B)
print("Matrix mul", C)


D = A @ B
print("Matrix mul", D)

```

```python
# Simple Neural Net Module

class SimpleNN(nn.Module):
  def __init__(self):
    super().__init__()
    # define layer
    self.layer = nn.Linear(5,3) # 5 inputs/ 3 o/p

  def forward(self,x):
    return self.layer(x)
model = SimpleNN()
print(model)
```

```python
x = torch.randn(2, 5)  # Batch of 2, 5 features each
output = model(x)
print("\nInput shape:", x.shape)
print("Output shape:", output.shape)
```

```python
# Linear layer: y = Wx + b
linear = nn.Linear(in_features=10, out_features=5)
x = torch.randn(3, 10)  # Batch of 3
output = linear(x)
print("Linear layer output:", output.shape)  # torch.Size([3, 5])

# ReLU activation: max(0, x)
relu = nn.ReLU()
x = torch.tensor([-2.0, -1.0, 0.0, 1.0, 2.0])
output = relu(x)
print("\nReLU output:", output)  # tensor([0., 0., 0., 1., 2.])

# Combining them
x = torch.randn(2, 10)
x = linear(x)
print("Before ReLU:", x)
x = relu(x)
print("After ReLU:", x)  # All negative values → 0
```

```python
class ThreeLayerNetwork(nn.Module):
  def __init__(self):
    super().__init__()

    # input layer
    self.layer1 = nn.Linear(10,20) # ip 10 .hidden 20
    self.relu1 = nn.ReLU()

    self.layer2 = nn.Linear(20,20)
    self.relu2 = nn.ReLU()

    self.layer3 = nn.Linear(20,1)


  def forward(self,x):
    x = self.layer1(x)
    x = self.relu1(x)

    x = self.layer2(x)
    x = self.relu2(x)

    return x
```

```python
model = ThreeLayerNetwork()
print(model)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
print(f"\nTotal parameters: {total_params}")

# Forward pass with random data
batch_size = 5
x = torch.randn(batch_size, 10)  # 5 samples, 10 features each
output = model(x)

print(f"\nInput shape: {x.shape}")
print(f"Output shape: {output.shape}")
print(f"Output:\n{output}")
```

```python
class CompactNetwork(nn.Module):
  def __init__(self):
    super().__init__()

    # input layer

    self.network = nn.Sequential(
    nn.Linear(10,20), # ip 10 .hidden 20
    nn.ReLU(),
    nn.Linear(20,20),
    nn.ReLU(),
    nn.Linear(20,1),
    )


  def forward(self,x):

    return self.network(x)
```

```python
model2 = CompactNetwork()
output2 = model2(torch.randn(5, 10))
print("Compact network output shape:", output2.shape)
```
