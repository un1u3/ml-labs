```python
import torch
import torch.nn as nn
```

```python
# creating a tensor
x = torch.tensor([1,2,3,4,5])
print(x)

# 2d tensors
y = torch.tensor([[1,2,3],
                  [4,5,6]])
print(y)

# creating tensors with zeros
z = torch.zeros([3,3])
print(z)

# with random
a = torch.randn(6,7)
print(a)

# with ones
torch.ones(2,3)
```

```python
# tensor operations
x = torch.randn(2,3,4)
print("Original Shape",x.shape)

y = x.reshape(6,4)
print("ReShape",y.shape)

z  = y.view(-1)
print("Flatten",z.shape)
```

```python
#  Common operations
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print("\nAddition:", a + b)
print("Multiplication:", a * b)  # element-wise
print("Mean:", a.mean())
print("Sum:", a.sum())
```

```python
# Matrix multiplication (dot product)
A = torch.tensor([[1, 2],
                  [3, 4],
                  [5, 6]])  # 3x2

B = torch.tensor([[7, 8, 9],
                  [10, 11, 12]])  # 2x3


C = torch.matmul(A,B)
print("Matrix mul", C)


D = A @ B
print("Matrix mul", D)

```

```python
# autograd

x = torch.tensor(4.0,requires_grad = True)
x
```

```python
y = x ** 2
```

```python
y.backward()
```

```python
print("dy/dx",x.grad)
```

```python
# more complex differentatio
x = torch.tensor(20.0,requires_grad=True)
y = x ** 2 + x + x ** 2
y.backward()
x.grad
```
