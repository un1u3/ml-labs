```python
# Tensors
# A tensor is a generalization of vector and matrices
# Scalar = 0-D tensor
# Vecotr = 1-D tendor
# Matrix = 2-D tensor
# high dimensional tensor = 3-D, 4D , nD
# tensors are similar to matrirces and NumPy's nDarray
# tensor can run in GPU
```

```python
import torch
import numpy as np
```

```python
# tensor initialization
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
x_data
```

```python
# more on tensors
print(torch.rand(3,4))
print(torch.ones(3,2))
print(torch.zeros(3,4))
```

```python
# tensor attribute
tensor = torch.rand(3,4)
print(tensor.shape)
print(tensor.dtype)
print(tensor.device)
```

```python
# tensor operation
# 1. element wise operation
x = torch.tensor([1,2,3])
y = torch.tensor([4,5,6])
print( x + y )
print( x - y )
print( x * y )
print( x / y )


```

```python
# 2.matrix multiplication/Dot product
A = torch.tensor([[1,2],[3,4]])
B = torch.tensor([[5,6],[7,8]])
torch.matmul(A,B)
```
