```python
data = torch.tensor([
    [2, 1],
    [3, 2],
    [4, 2],
    [5, 3],
    [1, 1],
    [6, 4],
    [7, 5],
    [3, 3],
    [8, 6],
    [2, 2],
    [9, 7],
    [4, 3],
    [5, 4],
    [6, 5],
    [7, 6]
],dtype=torch.float32)
y = torch.tensor([30,40,45,55,25,65,75,50,85,35,90,48,60,70,80], dtype=torch.float32).view(-1,1)
```

```python
model = nn.Sequential(
    nn.Linear(2,8), # 2 ip 
    nn.ReLU(),
    nn.Linear(8,1) # 1 op 
)
```

```python
import torch
import torch.nn as nn

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
for epoch in range(500):
    output = model(data)
    loss = criterion(output, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
print("final loss",loss.item())
```
