```python
import numpy as np 

def conv2d_single_channel(img, kernel, stride = 1, padding = 0):
    H, W = img.shape
    K = kernel.shape[0]



    out_h = (H - K) + 1 
    out_w = (W-K) + 1
    out = np.zeros((out_h, out_w))


    for i in range(out_h):
        for j in range(out_w):
            h_start, w_start = i * stride, j * stride
            h_end, w_end = h_start + K, w_start + K
            
            patch = img[h_start:h_end, w_start:w_end]
            out[i, j] = np.sum(patch * kernel)
    return out 

    
```
