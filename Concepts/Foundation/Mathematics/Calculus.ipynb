```python
import torch 
import numpy as np 
from matplotlib_inline import backend_inline 
```

```python
# derivative is the rate of change in a function with respect ot chnage to its arguments 
# derivatives can tell us how rapidly a loss function would increase or decreaes when we increase or 
# decrease each parameter by a infinitesimally small amount
```

```python
def f(x):
    return 3* x**2 -4 * x

for h in 10.0**np.arange(-1,-6,-1):
    print(f'h={h:.5f}, numerical limit = {(f(1+h)-f(1))/h}:.5f')
```

```python
# automatic differentaion 

x = torch.arange(4.0, requires_grad=True)
x
```

```python
x.grad
```

```python
y = 2 * torch.dot(x,x)
y
```

```python
y.backward()
```

```python
x.grad == 4 * x
```

```python
# reset the gradients 
x.grad.zero_()
y = x.sum()
y.backward()
x.grad()
```
