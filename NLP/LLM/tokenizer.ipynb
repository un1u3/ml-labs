```python
all_words = sorted(set(text))
len(all_words)
```

```python
class Tokenizer:

    def __init__(self,text):
        self.chars = sorted(list(set(text)))
        self.vocab_size = len(self.chars)

        self.stoi = {}
        for i,char in enumerate(self.chars):
            self.stoi[char] = i

        self.itos = {}
        for i,char in enumerate(self.chars):
            self.itos[i] = char
    
    def encode(self,text):
        indices = []
        for c in text:
            indices.append(self.stoi[c])
        return indices

    def decode(self,indices):
        chars = []
        for char in indices:
            chars.append(self.itos[char])
        return ''.join(chars)



        
```

```python
Tk = Tokenizer(rawtext)
print(Tk.chars)
print(Tk.vocab_size)
print(Tk.itos)
print(Tk.encode("hello my name is unique "))
Tk.decode([49, 46, 53, 53, 56, 0, 54, 66, 0, 55, 42, 54, 46, 0, 50, 60, 0, 62, 55, 50, 58, 62, 46, 0])
```
