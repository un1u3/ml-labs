# 1. Text Cleaning

```python
# eg
corpus = [
    "My laptop is heating like a mini nuclear reactor during online classes.",
    "Scientists just detected water vapor on a planet 120 light-years away.",
    "The Nepali rupee strengthened slightly against the U.S. dollar today.",
    "<div class='post'>Error 404: Knowledge not found.</div>",
    "C++ templates make my brain twist like a pretzel???!!!",
    "AI models are getting better at understanding images, speech, and even music.",
    "I overslept again... someone please invent an alarm clock that argues with me.",
    "Researchers released a dataset with 40 million protein structures last week.",
    "Bitcoin dropped sharply after unexpected market volatility.",
    "Why does my code run perfectly at 2 AM but throws errors at 2 PM??"
]

```

```python
# 1.Text cleaning
# we will convert the text to lowercas, remove puncuation,numbers speccial characters and HTML tags

import re
import string
from bs4 import BeautifulSoup

def clean_text(text):
  text.lower() # lower case
  text = re.sub(r'\d','', text) #remove numbers
  text = text.translate(str.maketrans('','',string.punctuation)) # remove puncutaion
  text = re.sub(r'\W', ' ',text) # remove specail character
  text = BeautifulSoup(text,'html.parser').get_text() # removes html tags
  return text

```

```python
cleaned_corpus = [clean_text(doc) for doc in corpus]
```

```python
cleaned_corpus
```

# 2. Tokenization
- splitting the cleaned data into tokens(words)

```python
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')

tokenize_corpus = [word_tokenize(doc) for doc in cleaned_corpus]
```
