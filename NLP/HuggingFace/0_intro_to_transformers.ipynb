```python
from transformers import pipeline
classifer = pipeline('sentiment-analysis')
classifer("Ok I am doing NLPPP")
```

```python
classifer(["I am very happy","I am very  sad"])
```

```python
classifer = pipeline('zero-shot-classification')
classifer("this is an Animal",
    candidate_labels = ['Animal','education','man','Tiger','women','moon','neil armstrong'])
```

```python
generator = pipeline('text-generation')
generator('I am here today to study about AI ')
```

```python
unmasker = pipeline('fill-mask')
unmasker('This will teach you about <mask> model')
```

```python
question_answer = pipeline("question-answering")
question_answer(

    question ="where do I work?",
    context = 'My name is unique and I work at my company'
)
```
