```python
from transformers import AutoTokenizer 

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

sequence = "I;vs been wating here"
modelInputs = tokenizer(sequence)
modelInputs


# what is happening here 
# 1. Tokenizes the text into subwords 
# 2. Converts token to input IDs
# 3. Adds special tokens 
# 4. creates attention masks
# 5.Returns everything in a dictionary format 
```

```python
# handeling single vs Multiple sequences 

sequence = "I have been waiting for this"
model_input = tokenizer(sequence)
model_input
```

```python
# multiple input 
sequence = ["I've been waiting for a HuggingFace course my whole life.", 
             "So have I"]
model_inputs = tokenizer(sequence)
model_inputs
```
