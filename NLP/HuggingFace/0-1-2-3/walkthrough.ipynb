````python
from transformers import AutoTokenizer 

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# tokenize 
tokens = tokenizer.tokenize("natural langauge processing")
# ['natural', 'lang', '##au', '##ge', 'processing']
token_ids = tokenizer.convert_tokens_to_ids(tokens)
# o/p : [3019, 11374, 4887, 3351, 6364]

encoding = tokenizer("Natural Langauge Processing",return_tensors="pt")
encoding
````

### Step 2: Create Embedings 
convert token IDs intto dense vector 
` token ID 3019("natural") -> dense vector [0.2, -0.5, 0.8, ...]`

similar tokens have similar vectors 

Positional encoding
- add position information so model knows word order 
- Embeddings are learned during training
- After training, similar words have similar embeddings:

- Dog embedding ≈ Puppy embedding (semantically similar)
- Good embedding ≈ Great embedding (synonyms)
- Man embedding - Woman embedding ≈ King embedding - Queen embedding

### Step 3: Self Attention 
the core mechanism. for each token, calculate:
```
1. Query (Q): "What am I looking for?"
2. Key (K): "What information do I have?"
3. Value (V): "What information should I pass on?"

Attention Weight = softmax(Q · K^T / √d_k)
Output = Attention Weight · V
```

````python
# Query for "bank": "I'm a financial institution"
Q_bank = [0.8, 0.2, -0.1, ...]

# Keys in sentence:
K_the = [0.1, 0.5, 0.2, ...]    (less relevant)
K_bank = [0.9, 0.1, 0.3, ...]   (very relevant - itself)
K_executive = [0.7, 0.8, 0.1, ...] (relevant - describes type)

# Attention weights:
attention_the =~ 0.1
attention_bank =~ 0.3
attention_executive =~0.6

# Output = weighted sum of Values
output_bank = 0.1*V_the + 0.3*V_bank + 0.6*V_executive
````

### Step 4: Multi-Head Attention 

instead of one attention mechanism, using muliple in parallel 

- Head 1: Focuses on grammatical relationships 
- Head 2: Focuses on semantic relationships 
- Head 3: Focuses on entity references 
- Head 4: Focuses on co-reference 
...........
- Result: Concatenate all heads for rich representation
